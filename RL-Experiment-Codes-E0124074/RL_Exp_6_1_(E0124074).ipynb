### 3. Explore different environments in RL and define its MDP

Let's explore a classic control environment, 'CartPole-v1', to understand its MDP components. The MDP (Markov Decision Process) is typically defined by its states, actions, transition function, and reward function.

```python
# Load the environment (e.g., CartPole-v1)
env = gym.make('CartPole-v1')

print(f"\nEnvironment: {env.spec.id}")
print(f"\nAction Space: {env.action_space}")
if isinstance(env.action_space, gym.spaces.Discrete):
    print(f"Number of discrete actions: {env.action_space.n}")
print(f"\nObservation Space: {env.observation_space}")
if isinstance(env.observation_space, gym.spaces.Box):
    print(f"Observation Space Shape: {env.observation_space.shape}")
    print(f"Observation Space Low: {env.observation_space.low}")
    print(f"Observation Space High: {env.observation_space.high}")
```
